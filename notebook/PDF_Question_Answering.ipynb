# PDF Question Answering System using LangChain

## Domain: AI in Education

### Project Overview

This project implements a PDF-based Question Answering System using LangChain, ChromaDB, Sentence Transformers, and Ollama.

The system extracts text from a PDF document, splits the text into smaller chunks, generates embeddings, and retrieves the top 5 relevant chunks for each question.

The retrieved information is used by the Llama 3.2 model to generate accurate answers.

### Objectives

- Extract text from PDF documents.
- Split text into manageable chunks.
- Generate text embeddings.
- Store embeddings using ChromaDB.
- Retrieve relevant information for user questions.
- Generate answers using a local LLM.
- Compare answers with and without retrieval.

```python
import pandas as pd

print("Libraries imported successfully!")
```

```python
answers = pd.read_csv("../dataset/answers.csv")
chunks = pd.read_csv("../dataset/chunks.csv")

print("Datasets loaded successfully!")
print("Answers shape:", answers.shape)
print("Chunks shape:", chunks.shape)

display(answers.head())
```

```python
print("PDF QUESTION ANSWERING SYSTEM")
print("--------------------------------")

print("PDF Pages:", 15)
print("Total Chunks:", len(chunks))
print("Total Questions:", len(answers))
print("RAG Answers:", answers["Answer_With_Retrieval"].count())
print("Normal Answers:", answers["Answer_Without_Retrieval"].count())
```

```python
comparison = answers[
    [
        "Query_ID",
        "Question",
        "Answer_With_Retrieval",
        "Answer_Without_Retrieval"
    ]
]

display(comparison)
```

```python
retrieved_results = answers[
    ["Query_ID", "Question", "Retrieved_Chunks", "Retrieval_Scores"]
]

display(retrieved_results.head(5))
```

## Conclusion

The PDF Question Answering System successfully processes a 15-page educational PDF using LangChain.

The system extracts text, creates 71 chunks, and generates 384-dimensional embeddings using Sentence Transformers. ChromaDB stores the embeddings and retrieves the top 5 relevant chunks for each question.

The Llama 3.2 model generates answers using Ollama. The project also compares answers generated with retrieval and without retrieval.

This project demonstrates how Retrieval-Augmented Generation (RAG) can be used to build an intelligent PDF-based Question Answering System.
